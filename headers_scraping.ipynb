```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from tqdm import tqdm
import json
```

```python
headers = {
        'User-Agent': 'Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Version/4.0 Chrome/130.0.6723.58 Safari/537.36 (AirWatch Browser v21.09.0.9)'
        }
```

```python
url='https://explore.whatismybrowser.com/useragents/explore/software_type_specific/web-browser/'
```

```python
r = requests.get(url,headers=headers, timeout=10)
print(str(r.status_code))
soup = BeautifulSoup(r.content)
```

```python
headers_dict = {}
```

```python
for i, entry in enumerate(tqdm(soup.find_all('a', class_='code'))):
    headers_dict[i] = entry.text
```

```python
headers_dict
```

```python
with open("headers.json", "w") as f:
    json.dump(headers_dict, f)
```
